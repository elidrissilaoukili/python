```python
import pandas as pd
from bs4 import BeautifulSoup
```

```python
with open('./file.html', 'r') as f:
    html = f.read()
soup = BeautifulSoup(html, 'html.parser')
```

```python
data = []
rows = soup.find_all('tr')
```

```python
for row in rows:
    if(row.find('td', class_='text')):
        question = row.find('td', class_='text').get_text(strip=True)
    if(row.find('select')):
        options = row.find('select').find_all('option')
    choices = [opt.get_text(strip=True) for opt in options if opt['value'] != '0']
    data.append({"Question": question, "Options": choices})
    data.append({})
```

```python
# Convert to DataFrame
df = pd.DataFrame(data)
```

```python
df.to_csv('extracted_data.csv', index=False)
```
