```python
import pandas as pd
from datetime import datetime
```

```python
df = pd.read_csv('./MLAIM.csv')
```

```python
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
df['birth_date'] = pd.to_datetime(df['birth_date'], format='mixed')
```

```python
today = datetime.today()
df['Age'] = df['birth_date'].apply(lambda x: today.year - x.year - ((today.month, today.day) < (x.month, x.day)))
get_birth_date_index = df.columns.get_loc('birth_date')
df.insert(get_birth_date_index + 1, "Age", df.pop('Age'))
```

```python
firstName = df['firstName'].str.lower().str.replace(' ', '')
lastName = df['lastName'].str.lower().str.replace(' ', '')
df['Email'] = firstName + "." + lastName + "@usmba.ac.ma"
get_firstName_index = df.columns.get_loc('firstName')
df.insert(get_firstName_index + 1, "Email", df.pop('Email'))
```

```python
df['Accepted'] = ((df['Age'] < 21) & (df['birth_city'].isin(['FES', 'MEKNES']))).map({
    True: "Yes",
    False: 'No'
})
```

```python
df.to_csv('cleaned_data.csv', index=False)
```
